```python
from langchain.embeddings import GooglePalmEmbeddings
from langchain.llms import GooglePalm

llm = GooglePalm(google_api_key='YOUR_PALM2_API_KEY',
                 temperature=0.5,
                top_k=1)
```

### ConversationBufferMemory

```python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory
```

```python
memory = ConversationBufferMemory()
```

### Plugging in the memory to the conversation

```python
chat_conversation = ConversationChain(llm=llm,
                                      memory=memory,
                                      verbose=True)
```

```python
chat_conversation.predict(input='What do you know about AI')
```

```python
chat_conversation.predict(input='What do you know about Llama2?')
```

```python
print(memory.buffer)
```

```python
print(memory.load_memory_variables({}))
```

```python
import pickle
pickled = pickle.dumps(chat_conversation.memory)
```

### ConversationBufferWindowMemory

```python
from langchain.memory import ConversationBufferWindowMemory

memory_1 = ConversationBufferWindowMemory(k=1)

chat_conversation_window = ConversationChain(llm=llm,
                                            memory=memory_1,
                                            verbose=True)
```

```python
chat_conversation_window.predict(input='Hello, How are you doing today?')
```

```python
chat_conversation_window.predict(input="What's the date today?")
```

```python
print(memory_1.buffer)
```

### ConversationSummaryBufferMemory

```python
from langchain.memory import ConversationSummaryBufferMemory

memory_summ = ConversationSummaryBufferMemory(llm=llm,
                                            max_token_limit=100)

chat_conversation_summ = ConversationChain(llm=llm,
                                            memory=memory_summ,
                                            verbose=True)
```

```python
chat_conversation_summ.predict(input='What do you know about AI?')
```

```python
chat_conversation_summ.predict(input='What do you know about Machine Learning?')
```

```python
memory_summ.load_memory_variables({})
```
